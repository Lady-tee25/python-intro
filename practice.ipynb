```python
#<-- Calculating the discount amount and the final price of a Product.


product_price = 250.00      #<-- PRODUCT PRICE, 250.00
discount_rate = 0.15        #<-- DISCOUNT RATE, 0.15
discount_amount = product_price * discount_rate     #<-- FINDING DISCOUNT AMOUNT
final_price = product_price - discount_amount       #<-- CALCULATING THE FINAL PRICE

print("Original price:", product_price)
print("Discount:", discount_amount)
print("Final price:", final_price)
```
