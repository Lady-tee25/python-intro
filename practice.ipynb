```python
Inventory = [
    ("rice", 5000, 3),
    ("beans", 3500, 8),
    ("garri", 2000, 5),
    ("yam", 5000, 4) 
]

Inventory.append(
    ("bread", 4000, 6)
)
Inventory.append(
    ("milk", 8000, 10)
)
print(Inventory)

#To remove a product
for product in Inventory: 
    name, price, quantity = product
    if product[1] == "beans":
        Inventory.remove(product)

print(Inventory)

#Enumerated inventory
for num, product in enumerate(Inventory, start=1):
    print(f"{num} - {product}")

#Total amount
total_price = 0

for product in Inventory:
    name, price, quantity = product
    total_price += product[1] * product[2]

print(f"total inventory price = {total_price}")


#Most expensive product
most_expensive = Inventory[2]

for product in Inventory:
    if product[1] > most_expensive[2]:
        most_expensive = product

print(most_expensive[0:-1])
```

```python
#Course enrollment tracker

enrollment = {
    "python" : {"Ada", "Bola", "Tolu"},
    "ai" : {"Bola", "mercy", "fola"}
}

#add 2 new students
enrollment["python"].add("philip")
enrollment["python"].add("alice")

#get intersection and difference
student_in_both = enrollment["python"].intersection(enrollment["ai"])
only_python = enrollment["python"].difference(enrollment["ai"])


print(enrollment)
print(student_in_both)
print(only_python)

print("=" * 20)
print(  "solution 4")
print("=" * 20)

for course, students in enrollment.items():
    print(f"{course} - {len(students)}")


#finding the course with the highest student count
highest_count = 0
winning_course = ""

for course, student in enrollment.items():
    if len(student) > highest_count:
        winning_course = course

print(winning_course)
```
